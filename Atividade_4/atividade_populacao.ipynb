# População brasileira: 2010 e 2022

Agregações por estado e município, ordenadas pela variação absoluta da população entre 2010 e 2022.

```python
import pandas as pd

entrada = 'cd2022_populacao2010.xlsx'

dados = pd.read_excel(entrada, sheet_name=0, skiprows=2)
dados = dados[dados['COD. UF'].notna()].copy()
dados.columns = ['IGNORAR', 'UF', 'COD_UF', 'COD_MUNIC', 'MUNICIPIO', 'POP_2010_SINOPSE', 'POP_2010_COMPATIBILIZADA', 'POP_2022']

col_2010 = 'POP_2010_COMPATIBILIZADA'
col_2022 = 'POP_2022'
dados['Crescimento 2010-2022'] = dados[col_2022] - dados[col_2010]

dados.head()
```

```python
p_estado = (
    dados.groupby('UF')[[col_2010, col_2022, 'Crescimento 2010-2022']]
      .sum()
      .reset_index()
      .sort_values('Crescimento 2010-2022', ascending=False)
)

p_estado.to_csv('populacao_estado.csv', sep=';', index=False, encoding='utf-8-sig')
p_estado
```

```python
p_municipio = (
    dados[['UF', 'COD_UF', 'COD_MUNIC', 'MUNICIPIO', col_2010, col_2022, 'Crescimento 2010-2022']]
      .sort_values('Crescimento 2010-2022', ascending=False)
)

p_municipio.to_csv('populacao_municipio.csv', sep=';', index=False, encoding='utf-8-sig')
p_municipio.head()
```
